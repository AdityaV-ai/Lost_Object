```python
import torch
import open_clip

device = "cuda" if torch.cuda.is_available() else "cpu"

print("PyTorch version:", torch.__version__)
print("GPU available:", torch.cuda.is_available())
print("Using device:", device)
```

```python
import torch
import open_clip

model, _, preprocess = open_clip.create_model_and_transforms(
    "ViT-B-32",
    pretrained="laion2b_s34b_b79k"
)

tokenizer = open_clip.get_tokenizer("ViT-B-32")

model = model.to(device)
model.eval()

print("OpenCLIP ViT-B/32 loaded successfully!")
print("Device:", device)
```

```python
import os

dataset_path = "/content/drive/MyDrive/Dataset"
images_path = os.path.join(dataset_path, "Images80+")

image_files = [
    f for f in os.listdir(images_path)
    if f.lower().endswith((".jpg", ".jpeg", ".png"))
]

print("Total images:", len(image_files))
print("First 10 images:", image_files[:10])
```

```python
from PIL import Image
import torch
import os

image_embeddings = {}
failed_images = []

for filename in image_files:
    image_path = os.path.join(images_path, filename)

    try:
        image = Image.open(image_path).convert("RGB")
        image_tensor = preprocess(image).unsqueeze(0).to(device)

        with torch.no_grad():
            embedding = model.encode_image(image_tensor)
            embedding = embedding / embedding.norm(dim=-1, keepdim=True)

        image_embeddings[os.path.splitext(filename)[0]] = embedding.cpu()

    except Exception as e:
        failed_images.append((filename, str(e)))

print("Total embeddings created:", len(image_embeddings))
print("Failed images:", failed_images)
print("Embedding shape:", next(iter(image_embeddings.values())).shape)
```

```python
import pandas as pd

gt_path = "/content/drive/MyDrive/Dataset/ground_truth_pairs.csv"
gt = pd.read_csv(gt_path)

print("Total ground-truth pairs:", len(gt))
print(gt.head())
```

```python
import torch
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity

results = []

for _, row in gt.iterrows():
    lost_id = row["lost_report_id"]
    found_id = row["found_report_id"]

    lost_embedding = image_embeddings[lost_id]
    found_embedding = image_embeddings[found_id]

    similarity = cosine_similarity(
        lost_embedding.numpy(),
        found_embedding.numpy()
    )[0][0]

    results.append({
        "pair_id": row["pair_id"],
        "lost_report_id": lost_id,
        "found_report_id": found_id,
        "label": row["label"],
        "image_similarity": similarity
    })

similarity_df = pd.DataFrame(results)

print("Total pairs evaluated:", len(similarity_df))
print("\nFirst 10 results:")
display(similarity_df.head(10))
```

```python
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    accuracy_score,
    balanced_accuracy_score
)

y_true = similarity_df["label"]
y_scores = similarity_df["image_similarity"]

# Basic evaluation
auc = roc_auc_score(y_true, y_scores)
ap = average_precision_score(y_true, y_scores)

# Initial threshold for exploration
y_pred = (y_scores >= 0.5).astype(int)

accuracy = accuracy_score(y_true, y_pred)
balanced_acc = balanced_accuracy_score(y_true, y_pred)

print("OPENCLIP IMAGE SIMILARITY EVALUATION")
print("-----------------------------------")
print("Total pairs:", len(y_true))
print("ROC-AUC:", round(auc, 4))
print("Average Precision:", round(ap, 4))
print("Accuracy at threshold 0.5:", round(accuracy, 4))
print("Balanced Accuracy:", round(balanced_acc, 4))

print("\nAverage similarity by class:")
print(similarity_df.groupby("label")["image_similarity"].mean())
```

```python
import numpy as np
from sklearn.metrics import balanced_accuracy_score

thresholds = np.arange(0.30, 0.91, 0.01)

threshold_results = []

for threshold in thresholds:
    predictions = (
        similarity_df["image_similarity"] >= threshold
    ).astype(int)

    score = balanced_accuracy_score(
        similarity_df["label"],
        predictions
    )

    threshold_results.append((threshold, score))

best_threshold, best_score = max(
    threshold_results,
    key=lambda x: x[1]
)

print("Best threshold:", round(best_threshold, 2))
print("Best balanced accuracy:", round(best_score, 4))
```

```python
import pandas as pd

items_df = pd.read_csv("items_updated - items_updated.csv")

print("Total reports:", len(items_df))
print("Lost reports:", len(items_df[items_df["status"] == "Lost"]))
print("Found reports:", len(items_df[items_df["status"] == "Found"]))

print("\nLost item descriptions:")
display(
    items_df[items_df["status"] == "Lost"][
        ["report_id", "category", "description"]
    ].head(5)
)
```

```python
lost_df = items_df[
    (items_df["status"] == "Lost") &
    (items_df["description"].notna())
].copy()

text_embeddings = {}

for _, row in lost_df.iterrows():
    report_id = row["report_id"]
    description = str(row["description"])

    tokens = tokenizer([description]).to(device)

    with torch.no_grad():
        embedding = model.encode_text(tokens)
        embedding = embedding / embedding.norm(dim=-1, keepdim=True)

    text_embeddings[report_id] = embedding.cpu()

print("Text embeddings created:", len(text_embeddings))
print("Embedding shape:", next(iter(text_embeddings.values())).shape)
```

```python
import pandas as pd
import torch

cross_modal_results = []

for _, row in items_df[items_df["status"] == "Found"].iterrows():

    found_id = row["report_id"]
    found_embedding = image_embeddings[found_id]

    for lost_id, text_embedding in text_embeddings.items():

        similarity = torch.sum(
            found_embedding * text_embedding
        ).item()

        cross_modal_results.append({
            "found_report_id": found_id,
            "lost_report_id": lost_id,
            "image_text_similarity": similarity
        })

cross_modal_df = pd.DataFrame(cross_modal_results)

print("Total image-text comparisons:", len(cross_modal_df))
print("\nTop 10 matches:")
display(
    cross_modal_df.sort_values(
        "image_text_similarity",
        ascending=False
    ).head(10)
)
```

```python
# Join image-text results with ground-truth labels
evaluation_df = cross_modal_df.merge(
    gt[["lost_report_id", "found_report_id", "label"]],
    on=["lost_report_id", "found_report_id"],
    how="left"
)

print("Total comparisons:", len(evaluation_df))
print("Pairs with ground-truth labels:",
      evaluation_df["label"].notna().sum())

print("\nTop 10 results with labels:")

display(
    evaluation_df.sort_values(
        "image_text_similarity",
        ascending=False
    ).head(10)
)
```

```python
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    accuracy_score,
    balanced_accuracy_score
)

# Keep only pairs with known ground-truth labels
labeled_df = evaluation_df.dropna(subset=["label"]).copy()

y_true = labeled_df["label"].astype(int)
y_scores = labeled_df["image_text_similarity"]

# Evaluate ranking performance
auc = roc_auc_score(y_true, y_scores)
ap = average_precision_score(y_true, y_scores)

# Initial threshold
y_pred = (y_scores >= 0.25).astype(int)

print("OPENCLIP IMAGE-TO-TEXT EVALUATION")
print("--------------------------------")
print("Labelled pairs:", len(labeled_df))
print("ROC-AUC:", round(auc, 4))
print("Average Precision:", round(ap, 4))
print("Accuracy:", round(accuracy_score(y_true, y_pred), 4))
print("Balanced Accuracy:", round(
    balanced_accuracy_score(y_true, y_pred), 4
))

print("\nAverage similarity by class:")
print(labeled_df.groupby("label")["image_text_similarity"].mean())
```

```python
all_pair_results = []

lost_ids = [
    rid for rid in items_df[
        (items_df["status"] == "Lost")
    ]["report_id"]
    if rid in image_embeddings
]

found_ids = [
    rid for rid in items_df[
        (items_df["status"] == "Found")
    ]["report_id"]
    if rid in image_embeddings
]

for found_id in found_ids:
    for lost_id in lost_ids:

        similarity = torch.sum(
            image_embeddings[found_id] *
            image_embeddings[lost_id]
        ).item()

        all_pair_results.append({
            "lost_report_id": lost_id,
            "found_report_id": found_id,
            "image_similarity": similarity
        })

all_image_df = pd.DataFrame(all_pair_results)

print("Total combinations:", len(all_image_df))
print(all_image_df.head())
```

```python
# Merge both similarity results
all_fusion_df = all_image_df.merge(
    cross_modal_df,
    on=["lost_report_id", "found_report_id"],
    how="inner"
)

# Calculate combined similarity
all_fusion_df["combined_similarity"] = (
    0.5 * all_fusion_df["image_similarity"] +
    0.5 * all_fusion_df["image_text_similarity"]
)

print("Total fusion combinations:", len(all_fusion_df))

print(
    all_fusion_df.sort_values(
        by="combined_similarity",
        ascending=False
    ).head(10)
)
```

```python
# Add ground-truth labels to all fusion results
ranked_df = all_fusion_df.merge(
    gt[["lost_report_id", "found_report_id", "label"]],
    on=["lost_report_id", "found_report_id"],
    how="left"
)

# Sort by combined similarity
ranked_df = ranked_df.sort_values(
    by="combined_similarity",
    ascending=False
).reset_index(drop=True)

# Display top 10
top_10 = ranked_df.head(10)

print("TOP 10 MATCH RESULTS")
print(top_10[
    ["lost_report_id", "found_report_id",
     "combined_similarity", "label"]
])

print("\nCorrect matches in Top 10:",
      (top_10["label"] == 1).sum())

print("Incorrect matches in Top 10:",
      (top_10["label"] == 0).sum())
```
